```python
class EPO:
    def __init__(self, daily_path:Path, monthly_path:Path,date:str) -> None:
        self.daily_path = daily_path
        self.monthly_path = monthly_path
        self.date = date
        self.daily_price, self.monthly_price = self.take_training_data()
    def take_training_data (self, date:str, n_day:int, n_month:int):
        path_list: list[Path] = [self.daily_path, self.monthly_path]
        n_list: list[int] = [n_day,n_month]
        df_list: list[pd.DataFrame] = []
        for i in range(len(path_list)):
            df_temp: pd.DataFrame = pd.read_csv(path_list[i])
            df_temp['date'] = pd.to_datetime(df_temp['date'])
            df_temp.set_index(['ticker','date'],inplace=True)
            df_temp = df_temp['adj_close'].unstack(level=0)
            df_temp = df_temp.loc[:date].tail(n_list[i]+1)
            df_list.append(df_temp)
        return df_list
            
        
           
```
